### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
# Ingreso total por tienda
ingreso_tienda_1 = tienda["Precio"].sum()
ingreso_tienda_2 = tienda2["Precio"].sum()
ingreso_tienda_3 = tienda3["Precio"].sum()
ingreso_tienda_4 = tienda4["Precio"].sum()
```

```python
# Mostrar resultados
print(f"Ingreso total Tienda 1: ${ingreso_tienda_1:,.2f}")
print(f"Ingreso total Tienda 2: ${ingreso_tienda_2:,.2f}")
print(f"Ingreso total Tienda 3: ${ingreso_tienda_3:,.2f}")
print(f"Ingreso total Tienda 4: ${ingreso_tienda_4:,.2f}")
```

# 2. Ventas por categoría

```python
# Cantidad de productos vendidos por categoría en cada tienda

ventas_categoria_t1 = tienda.groupby("Categoría del Producto").size().sort_values(ascending=False)
ventas_categoria_t2 = tienda2.groupby("Categoría del Producto").size().sort_values(ascending=False)
ventas_categoria_t3 = tienda3.groupby("Categoría del Producto").size().sort_values(ascending=False)
ventas_categoria_t4 = tienda4.groupby("Categoría del Producto").size().sort_values(ascending=False)
```

```python
# Mostrar resultados
print("Tienda 1 - Ventas por categoría:")
print(ventas_categoria_t1, "\n")

print("Tienda 2 - Ventas por categoría:")
print(ventas_categoria_t2, "\n")

print("Tienda 3 - Ventas por categoría:")
print(ventas_categoria_t3, "\n")

print("Tienda 4 - Ventas por categoría:")
print(ventas_categoria_t4, "\n")
```

# 3. Calificación promedio de la tienda


```python
# Calificación promedio por tienda
promedio_t1 = tienda["Calificación"].mean()
promedio_t2 = tienda2["Calificación"].mean()
promedio_t3 = tienda3["Calificación"].mean()
promedio_t4 = tienda4["Calificación"].mean()
```

```python
# Mostrar resultados
print(f"Calificación promedio - Tienda 1: {promedio_t1:.2f}")
print(f"Calificación promedio - Tienda 2: {promedio_t2:.2f}")
print(f"Calificación promedio - Tienda 3: {promedio_t3:.2f}")
print(f"Calificación promedio - Tienda 4: {promedio_t4:.2f}")
```

# 4. Productos más y menos vendidos

```python
import matplotlib.pyplot as plt

def productos_mas_menos_vendidos(df, tienda_nombre):
    # Contar ventas por producto
    ventas_por_producto = df["Producto"].value_counts()
    # Mostrar el top 5 de productos más vendidos
    top_mas_vendidos = ventas_por_producto.head(5)
    top_menos_vendidos = ventas_por_producto.tail(5)

    print(f"\n{tienda_nombre} - Productos más vendidos:\n{top_mas_vendidos}")
    print(f"\n{tienda_nombre} - Productos menos vendidos:\n{top_menos_vendidos}")

    # Retornar las variables top_mas_vendidos y top_menos_vendidos
    return top_mas_vendidos, top_menos_vendidos

# Llamar a la función para cada tienda para obtener el top 5 y los últimos 5
top_mas_vendidos_t1, top_menos_vendidos_t1 = productos_mas_menos_vendidos(tienda, "Tienda 1")
top_mas_vendidos_t2, top_menos_vendidos_t2 = productos_mas_menos_vendidos(tienda2, "Tienda 2")
top_mas_vendidos_t3, top_menos_vendidos_t3 = productos_mas_menos_vendidos(tienda3, "Tienda 3")
top_mas_vendidos_t4, top_menos_vendidos_t4 = productos_mas_menos_vendidos(tienda4, "Tienda 4")


# Visualización para todas las tiendas
fig, axs = plt.subplots(4, 2, figsize=(14, 20))  # 4 filas para las 4 tiendas, 2 columnas para más y menos vendidos

# Tienda 1
top_mas_vendidos_t1.plot(kind="bar", ax=axs[0, 0], color="green")
axs[0, 0].set_title(f"Tienda 1 - Más vendidos")
axs[0, 0].set_ylabel("Cantidad vendida")
axs[0, 0].set_xlabel("Producto")

top_menos_vendidos_t1.plot(kind="bar", ax=axs[0, 1], color="red")
axs[0, 1].set_title(f"Tienda 1 - Menos vendidos")
axs[0, 1].set_ylabel("Cantidad vendida")
axs[0, 1].set_xlabel("Producto")

# Tienda 2
top_mas_vendidos_t2.plot(kind="bar", ax=axs[1, 0], color="green")
axs[1, 0].set_title(f"Tienda 2 - Más vendidos")
axs[1, 0].set_ylabel("Cantidad vendida")
axs[1, 0].set_xlabel("Producto")

top_menos_vendidos_t2.plot(kind="bar", ax=axs[1, 1], color="red")
axs[1, 1].set_title(f"Tienda 2 - Menos vendidos")
axs[1, 1].set_ylabel("Cantidad vendida")
axs[1, 1].set_xlabel("Producto")

# Tienda 3
top_mas_vendidos_t3.plot(kind="bar", ax=axs[2, 0], color="green")
axs[2, 0].set_title(f"Tienda 3 - Más vendidos")
axs[2, 0].set_ylabel("Cantidad vendida")
axs[2, 0].set_xlabel("Producto")

top_menos_vendidos_t3.plot(kind="bar", ax=axs[2, 1], color="red")
axs[2, 1].set_title(f"Tienda 3 - Menos vendidos")
axs[2, 1].set_ylabel("Cantidad vendida")
axs[2, 1].set_xlabel("Producto")

# Tienda 4
top_mas_vendidos_t4.plot(kind="bar", ax=axs[3, 0], color="green")
axs[3, 0].set_title(f"Tienda 4 - Más vendidos")
axs[3, 0].set_ylabel("Cantidad vendida")
axs[3, 0].set_xlabel("Producto")

top_menos_vendidos_t4.plot(kind="bar", ax=axs[3, 1], color="red")
axs[3, 1].set_title(f"Tienda 4 - Menos vendidos")
axs[3, 1].set_ylabel("Cantidad vendida")
axs[3, 1].set_xlabel("Producto")


plt.tight_layout()
plt.show()

```

Gráfico 3: Dispersión – Relación entre costo de envío y satisfacción del cliente

```python
promedios_envio = [tienda["Costo de envío"].mean(), tienda2["Costo de envío"].mean(), tienda3["Costo de envío"].mean(), tienda4["Costo de envío"].mean()]
promedios_nota = [tienda["Calificación"].mean(), tienda2["Calificación"].mean(), tienda3["Calificación"].mean(), tienda4["Calificación"].mean()]
tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

# Gráfico de dispersión
plt.figure(figsize=(8,6))
plt.scatter(promedios_envio, promedios_nota, color="purple", s=100)

# Anotar cada punto con el nombre de la tienda
for i, nombre in enumerate(tiendas):
    plt.text(promedios_envio[i]+0.02, promedios_nota[i], nombre, fontsize=9)

plt.title("Relación entre Costo de Envío y Satisfacción del Cliente")
plt.xlabel("Costo de envío promedio ($)")
plt.ylabel("Calificación promedio (nota)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

Costo de envío promedio

Tienda 2: ingresos sólidos y consistentes

Tienda 3: ingresos intermedios

Tienda 4: los ingresos más bajos entre todas

Tienda 4 se posiciona con alto costo de envío y buena calificación, pero sin reflejarlo en ventas

Las otras tiendas tienen un mejor balance entre costo y resultados

A pesar de tener una buena calificación por parte de los clientes, presenta las siguientes debilidades:

Ingresos significativamente más bajos que el resto de las tiendas

Menor variedad y volumen de productos vendidos

Costos logísticos elevados que afectan su rentabilidad

Poca participación en categorías clave del negocio

**Por estas razones, se recomienda al Sr. Juan vender la Tienda 4, con el objetivo de reducir costos, concentrar esfuerzos en tiendas más eficientes y maximizar el retorno de su negocio.**

Análisis Geográfico de Ventas por **Tienda**

Paso 1: Preparación de los datos

```python
import matplotlib.pyplot as plt

# Agregar una columna para identificar la tienda
tienda['tienda'] = 'Tienda 1'
tienda2['tienda'] = 'Tienda 2'
tienda3['tienda'] = 'Tienda 3'
tienda4['tienda'] = 'Tienda 4'

# Unificar todos los datos en un solo DataFrame
df_total = pd.concat([tienda, tienda2, tienda3, tienda4], ignore_index=True)
```

Paso 2: Gráfico de dispersión geográfico de ventas

```python
plt.figure(figsize=(12, 8))
colors = {'Tienda 1': 'red', 'Tienda 2': 'blue', 'Tienda 3': 'green', 'Tienda 4': 'purple'}

for tienda_id in df_total['tienda'].unique():
    subset = df_total[df_total['tienda'] == tienda_id]
    plt.scatter(subset['lon'], subset['lat'],
                label=tienda_id, alpha=0.4, s=20, c=colors[tienda_id])

plt.title("Distribución Geográfica de Ventas por Tienda")
plt.xlabel("Longitud")
plt.ylabel("Latitud")
plt.legend()
plt.grid(True)
plt.show()
```

Paso 3: Mapa de calor (Heatmap) con Folium

```python
!pip install folium
from folium import Map, plugins

# Crear mapa base
mapa = Map(location=[df_total['lat'].mean(), df_total['lon'].mean()], zoom_start=4)

# Crear heatmap con todas las ventas
heat_data = list(zip(df_total['lat'], df_total['lon']))
plugins.HeatMap(heat_data, radius=15).add_to(mapa)

mapa
```

Paso 4: Exploración de patrones regionales

```python
import seaborn as sns

# Crear un gráfico de dispersión con color según la calificación
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_total, x='lon', y='lat', hue='Calificación', palette='coolwarm', size='Precio', sizes=(10, 100))
plt.title("Calificaciones de Clientes por Ubicación")
plt.xlabel("Longitud")
plt.ylabel("Latitud")
plt.show()
```
